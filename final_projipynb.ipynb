```python

weather_df['preciptype'] = weather_df['preciptype'].astype('category').cat.codes
weather_df['conditions'] = weather_df['conditions'].astype('category').cat.codes
```

```python
from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()

weather_df["name_encoded"] = label_encoder.fit_transform(weather_df["name"])

print(dict(zip(label_encoder.classes_, label_encoder.transform(label_encoder.classes_))))
```

```python

import seaborn as sns
import matplotlib.pyplot as plt



features_to_check = ['precip', 'humidity', 'windspeed', 'windgust',
                     'dew', 'cloudcover', 'sealevelpressure',
                     'visibility', 'solarenergy']


def detect_outliers_iqr(weather_df, column):
    Q1 = weather_df[column].quantile(0.25)  
    Q3 = weather_df[column].quantile(0.75)  
    IQR = Q3 - Q1  
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = weather_df[(weather_df[column] < lower_bound) | (weather_df[column] > upper_bound)]
    return outliers

# Dictionary to store outlier counts
outlier_counts = {}

```

```python

# Detect and count outliers for each feature
for feature in features_to_check:
    outliers = detect_outliers_iqr(weather_df, feature)
    outlier_counts[feature] = len(outliers)
    print(f"Feature: {feature} | Outliers Detected: {len(outliers)}")

plt.figure(figsize=(15, 8))
for i, feature in enumerate(features_to_check, 1):
    plt.subplot(3, 3, i)
    sns.boxplot(x=weather_df[feature])
    plt.title(f'Boxplot for {feature}')
plt.tight_layout()
plt.show()
```

```python
# Apply log transformation for highly skewed features
weather_df['precip'] = np.log1p(weather_df['precip'])
weather_df['windgust'] = np.log1p(weather_df['windgust'])
weather_df['dew'] = np.log1p(weather_df['dew'])
weather_df['solarenergy'] = np.log1p(weather_df['solarenergy'])

# Remove extreme outliers using 1st-99th percentile for moderate impact features
for feature in ['humidity', 'windspeed', 'sealevelpressure']:
    lower_bound = weather_df[feature].quantile(0.01)
    upper_bound = weather_df[feature].quantile(0.99)
    weather_df = weather_df[(weather_df[feature] >= lower_bound) & (weather_df[feature] <= upper_bound)]

# Display final shape after outlier removal
print(f"Dataset shape after outlier handling: {weather_df.shape}")
```

```python

for feature in features_to_check:
    outliers = detect_outliers_iqr(weather_df, feature)
    outlier_counts[feature] = len(outliers)
    print(f"Feature: {feature} | Outliers Detected: {len(outliers)}")

plt.figure(figsize=(15, 8))
for i, feature in enumerate(features_to_check, 1):
    plt.subplot(3, 3, i)
    sns.boxplot(x=weather_df[feature])
    plt.title(f'Boxplot for {feature}')
plt.tight_layout()
plt.show()
```

```python
# Apply log transformation to precip & solarenergy to reduce skewness
weather_df['precip'] = np.log1p(weather_df['precip'])
weather_df['solarenergy'] = np.log1p(weather_df['solarenergy'])

# Adjust winsorization for dew (increase range)
lower_limit = weather_df['dew'].quantile(0.005)  
upper_limit = weather_df['dew'].quantile(0.995)  
weather_df['dew'] = weather_df['dew'].clip(lower=lower_limit, upper=upper_limit)

# Re-check outliers after transformation
outlier_counts_final = {}
for feature in features_to_check:
    outliers = detect_outliers_iqr(weather_df, feature)
    outlier_counts_final[feature] = len(outliers)
    print(f"Feature: {feature} | Outliers After Final Handling: {len(outliers)}")

# Final boxplots after adjustments
plt.figure(figsize=(15, 8))
for i, feature in enumerate(features_to_check, 1):
    plt.subplot(3, 3, i)
    sns.boxplot(x=weather_df[feature])
    plt.title(f'Final Boxplot for {feature}')
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

fsi_features = ['precip', 'humidity', 'windspeed', 'windgust', 'cloudcover', 'sealevelpressure']

scaler = MinMaxScaler()
weather_df[fsi_features] = scaler.fit_transform(weather_df[fsi_features])

# Adjusted weights to improve classification distribution
weights = {
    'precip': 0.55,          # Increased weight (main factor)
    'humidity': 0.25,        # Increased weight
    'windspeed': 0.08,       # Slightly reduced weight
    'windgust': 0.07,        # Slightly reduced weight
    'cloudcover': 0.03,      # Lower impact
    'sealevelpressure': 0.02 # Minimal impact
}

# Compute Flood Severity Index (FSI)
weather_df['FSI'] = (
    weights['precip'] * weather_df['precip'] +
    weights['humidity'] * weather_df['humidity'] +
    weights['windspeed'] * weather_df['windspeed'] +
    weights['windgust'] * weather_df['windgust'] +
    weights['cloudcover'] * weather_df['cloudcover'] +
    weights['sealevelpressure'] * weather_df['sealevelpressure']
) * 100  # Scale FSI to 100

# Adjusted flood severity thresholds
def classify_flood_severity(fsi):
    if fsi < 20:      
        return "No Flood"
    elif fsi < 40:    
        return "Low Flood"
    elif fsi < 65:    
        return "Moderate Flood"
    elif fsi < 85:
        return "High Flood"
    else:
        return "Severe Flood"

weather_df['Flood_Severity'] = weather_df['FSI'].apply(classify_flood_severity)

print(weather_df['Flood_Severity'].value_counts())

weather_df[['FSI', 'Flood_Severity']].head(10)
```

```python
# Capping (Winsorization) for Geographic
# Function to cap outliers using IQR method
def cap_outliers(df, columns):
    for col in columns:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR
        df[col] = df[col].clip(lower=lower_bound, upper=upper_bound)  # Capping outliers
    return df

# Apply outlier capping
cyclone_df = cap_outliers(cyclone_df, ['LAT', 'LON', 'DIST2LAND'])

print("Capping outliers completed for geographic.")
```

```python
import pandas as pd

# Assign new weights for available features
weight_speed = 0.5   # Storm speed is crucial for severity
weight_dir = 0.3     # Direction influences land impact
weight_distance = 0.2  # Distance to land affects strength

# Calculate Cyclone Severity Index (CSI)
cyclone_df['CSI_Score'] = (
    weight_speed * cyclone_df['STORM_SPEED'] +
    weight_dir * cyclone_df['STORM_DIR'] +
    weight_distance * (500 - cyclone_df['DIST2LAND'])  # Inverting distance (closer is worse)
)

# Calculate mean and standard deviation of CSI
mean_csi = cyclone_df['CSI_Score'].mean()
std_csi = cyclone_df['CSI_Score'].std()

# Define dynamic severity thresholds
no_cyclone_threshold = mean_csi - 1.5 * std_csi
low_threshold = mean_csi
moderate_threshold = mean_csi + 1 * std_csi
high_threshold = mean_csi + 2 * std_csi
severe_threshold = mean_csi + 3 * std_csi

# Classify severity dynamically
def classify_cyclone(csi):
    if csi <= no_cyclone_threshold:
        return 'No Cyclone'
    elif csi <= low_threshold:
        return 'Low Cyclone'
    elif csi <= moderate_threshold:
        return 'Moderate Cyclone'
    elif csi <= high_threshold:
        return 'High Cyclone'
    else:
        return 'Severe Cyclone'

# Apply classification
cyclone_df['CSI_Severity'] = cyclone_df['CSI_Score'].apply(classify_cyclone)

# Encode severity labels for machine learning
severity_mapping = {
    'No Cyclone': 0,
    'Low Cyclone': 1,
    'Moderate Cyclone': 2,
    'High Cyclone': 3,
    'Severe Cyclone': 4
}
cyclone_df['CSI_Severity_Encoded'] = cyclone_df['CSI_Severity'].map(severity_mapping)

# Display class distribution
print("\nCyclone Severity Distribution:\n", cyclone_df['CSI_Severity'].value_counts())
```

LSTM MODEL

```python
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report
import matplotlib.pyplot as plt
import seaborn as sns


# Ensure correct target column names
target_cyclone = "CSI_Severity_Encoded"
target_flood = "flood_severity_numeric"


# Features and Targets
features = ['precip', 'precipprob', 'precipcover', 'humidity', 'cloudcover',
            'windgust', 'windspeed', 'sealevelpressure', 'STORM_SPEED','STORM_DIR',
            'DIST2LAND', 'temp', 'feelslike']

X = df[features].values
y_cyclone = df[target_cyclone].values
y_flood = df[target_flood].values

# Check for Data Leakage
print("\n🔍 Checking for Missing Values:\n", df.isnull().sum())
print("\n🔍 Checking Data Distribution:\n", df[features].describe())

# Normalize Features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Reshape for LSTM (Samples, Timesteps, Features)
X_reshaped = X_scaled.reshape(X_scaled.shape[0], 1, X_scaled.shape[1])

# Train-Test Split
X_train, X_test, y_cyclone_train, y_cyclone_test, y_flood_train, y_flood_test = train_test_split(
    X_reshaped, y_cyclone, y_flood, test_size=0.2, random_state=42, stratify=y_cyclone)

# LSTM Model for Cyclone Prediction
model_cyclone = Sequential([
    LSTM(64, return_sequences=True, input_shape=(1, X_reshaped.shape[2])),
    Dropout(0.2),
    LSTM(32),
    Dropout(0.2),
    Dense(16, activation='relu'),
    Dense(len(np.unique(y_cyclone)), activation='softmax')  # Multiclass output
])

# Compile Cyclone Model
model_cyclone.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])

# Train Cyclone Model
history_cyclone = model_cyclone.fit(X_train, y_cyclone_train, validation_data=(X_test, y_cyclone_test),
                                    epochs=20, batch_size=32, verbose=1)

# Evaluate Cyclone Model
y_cyclone_pred = np.argmax(model_cyclone.predict(X_test), axis=1)
cyclone_acc_train = model_cyclone.evaluate(X_train, y_cyclone_train, verbose=0)[1]
cyclone_acc_test = accuracy_score(y_cyclone_test, y_cyclone_pred)

# LSTM Model for Flood Prediction
model_flood = Sequential([
    LSTM(64, return_sequences=True, input_shape=(1, X_reshaped.shape[2])),
    Dropout(0.2),
    LSTM(32),
    Dropout(0.2),
    Dense(16, activation='relu'),
    Dense(len(np.unique(y_flood)), activation='softmax')  # Multiclass output
])

# Compile Flood Model
model_flood.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])

# Train Flood Model
history_flood = model_flood.fit(X_train, y_flood_train, validation_data=(X_test, y_flood_test),
                                epochs=20, batch_size=32, verbose=1)

# Evaluate Flood Model
y_flood_pred = np.argmax(model_flood.predict(X_test), axis=1)
flood_acc_train = model_flood.evaluate(X_train, y_flood_train, verbose=0)[1]
flood_acc_test = accuracy_score(y_flood_test, y_flood_pred)

# Print Results
print(f"\n🚀 Cyclone Training Accuracy: {cyclone_acc_train:.4f}")
print(f"🚀 Cyclone Testing Accuracy: {cyclone_acc_test:.4f}")
print(f"🚀 Flood Training Accuracy: {flood_acc_train:.4f}")
print(f"🚀 Flood Testing Accuracy: {flood_acc_test:.4f}")

print("\nCyclone Severity Classification Report:\n", classification_report(y_cyclone_test, y_cyclone_pred))
print("\nFlood Severity Classification Report:\n", classification_report(y_flood_test, y_flood_pred))

```

LSTM + XGBOOST

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
import xgboost as xgb
from sklearn.metrics import accuracy_score, classification_report

# Define features and target variables
selected_features = [
    'precip', 'precipprob', 'precipcover',  # Rain-related
    'humidity', 'cloudcover', 'windgust', 'windspeed', 'sealevelpressure',  # Weather
    'STORM_SPEED', 'STORM_DIR', 'DIST2LAND',  # Storm-related
    'temp', 'feelslike'  # Temperature
]

Y_cyclone = df["CSI_Severity_Encoded"]
Y_flood = df["flood_severity_numeric"]

# Ensure selected features exist in the dataset
df_selected = df[selected_features]

# Split Data
X_train, X_test, Y_cyclone_train, Y_cyclone_test, Y_flood_train, Y_flood_test = train_test_split(
    df_selected, Y_cyclone, Y_flood, test_size=0.2, random_state=42, stratify=Y_cyclone
)

# Standardize Features
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Reshape for LSTM
X_train_lstm = X_train_scaled.reshape((X_train_scaled.shape[0], 1, X_train_scaled.shape[1]))
X_test_lstm = X_test_scaled.reshape((X_test_scaled.shape[0], 1, X_test_scaled.shape[1]))

# Build LSTM Model
lstm_model = Sequential([
    LSTM(64, activation='relu', return_sequences=True, input_shape=(1, X_train_scaled.shape[1])),
    Dropout(0.2),
    LSTM(32, activation='relu'),
    Dense(16, activation='relu'),
    Dense(2)  # Two outputs: Cyclone & Flood Severity
])

lstm_model.compile(optimizer='adam', loss='mse', metrics=['mae'])

# Train LSTM Model
lstm_model.fit(X_train_lstm, np.column_stack((Y_cyclone_train, Y_flood_train)),
                epochs=50, batch_size=16, validation_split=0.1, verbose=1)

# Extract LSTM Features
X_train_lstm_features = lstm_model.predict(X_train_lstm)
X_test_lstm_features = lstm_model.predict(X_test_lstm)

# Train XGBoost on LSTM Features
xgb_cyclone = xgb.XGBClassifier(objective='multi:softmax', num_class=4, eval_metric='mlogloss')
xgb_flood = xgb.XGBClassifier(objective='multi:softmax', num_class=5, eval_metric='mlogloss')

xgb_cyclone.fit(X_train_lstm_features, Y_cyclone_train)
xgb_flood.fit(X_train_lstm_features, Y_flood_train)

# Predictions
cyclone_preds = xgb_cyclone.predict(X_test_lstm_features)
flood_preds = xgb_flood.predict(X_test_lstm_features)

# Evaluation
print("Cyclone Accuracy:", accuracy_score(Y_cyclone_test, cyclone_preds))
print("Flood Accuracy:", accuracy_score(Y_flood_test, flood_preds))
print("\nCyclone Classification Report:")
print(classification_report(Y_cyclone_test, cyclone_preds))
print("\nFlood Classification Report:")
print(classification_report(Y_flood_test, flood_preds))

# Feature Importance Check
cyclone_importance = xgb_cyclone.feature_importances_
flood_importance = xgb_flood.feature_importances_
feature_names = [f'LSTM_Feature_{i}' for i in range(len(cyclone_importance))]
feature_importance_df = pd.DataFrame({
    'Feature': feature_names,
    'Cyclone_Importance': cyclone_importance,
    'Flood_Importance': flood_importance
})
print("\n🔍 Checking Feature Importance from XGBoost:")
print(feature_importance_df.sort_values(by='Cyclone_Importance', ascending=False))

# Data Leakage Check
duplicate_rows = pd.merge(X_train, X_test, how='inner').shape[0]
print("\n🔍 Checking for Possible Data Leakage...")
print(f"Number of identical rows between Train & Test: {duplicate_rows}")

# Target Distribution Check
def check_target_distribution(y_train, y_test, label):
    print(f"\n🔍 Checking Target Distribution Differences: {label}")
    train_dist = y_train.value_counts(normalize=True).rename('proportion')
    test_dist = y_test.value_counts(normalize=True).rename('proportion')
    print("Train Target Distribution:")
    print(train_dist)
    print("Test Target Distribution:")
    print(test_dist)

check_target_distribution(Y_cyclone_train, Y_cyclone_test, "Cyclone")
check_target_distribution(Y_flood_train, Y_flood_test, "Flood")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix, roc_curve, roc_auc_score

# Plot Confusion Matrix
def plot_confusion_matrix(y_true, y_pred, title):
    cm = confusion_matrix(y_true, y_pred)
    plt.figure(figsize=(10, 7))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=np.unique(y_true), yticklabels=np.unique(y_true))
    plt.xlabel('Predicted')
    plt.ylabel('Actual')
    plt.title(title)
    plt.show()

plot_confusion_matrix(Y_cyclone_test, cyclone_preds, 'Cyclone Confusion Matrix')
plot_confusion_matrix(Y_flood_test, flood_preds, 'Flood Confusion Matrix')

# Ensure history is being captured
history = lstm_model.fit(X_train_lstm, np.column_stack((Y_cyclone_train, Y_flood_train)),
                epochs=50, batch_size=16, validation_split=0.1, verbose=1)

# Plot Training & Testing Loss/Accuracy
def plot_training_history(history):
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

    # Loss
    ax1.plot(history.history['loss'], label='Train Loss')
    ax1.plot(history.history['val_loss'], label='Val Loss')
    ax1.set_title('Training & Validation Loss')
    ax1.set_xlabel('Epoch')
    ax1.set_ylabel('Loss')
    ax1.legend()

    # Accuracy
    ax2.plot(history.history['mae'], label='Train MAE')
    ax2.plot(history.history['val_mae'], label='Val MAE')
    ax2.set_title('Training & Validation MAE')
    ax2.set_xlabel('Epoch')
    ax2.set_ylabel('Mean Absolute Error')
    ax2.legend()

    plt.show()

plot_training_history(history)

# Plot ROC Curve in one graph
def plot_roc_curves(y_true, y_preds, title):
    plt.figure(figsize=(10, 7))
    for i in range(y_true.shape[1]):
        fpr, tpr, _ = roc_curve(y_true[:, i], y_preds[:, i])
        auc = roc_auc_score(y_true[:, i], y_preds[:, i])
        plt.plot(fpr, tpr, lw=2, label=f'Class {i} (AUC = {auc:.2f})')
    plt.plot([0, 1], [0, 1], color='gray', lw=2, linestyle='--')
    plt.xlim([0.0, 1.0])
    plt.ylim([0.0, 1.05])
    plt.xlabel('False Positive Rate')
    plt.ylabel('True Positive Rate')
    plt.title(title)
    plt.legend(loc="lower right")
    plt.show()

# Binarize the output for ROC curve
Y_cyclone_test_bin = pd.get_dummies(Y_cyclone_test).values
Y_flood_test_bin = pd.get_dummies(Y_flood_test).values
cyclone_preds_bin = pd.get_dummies(cyclone_preds).values
flood_preds_bin = pd.get_dummies(flood_preds).values

plot_roc_curves(Y_cyclone_test_bin, cyclone_preds_bin, 'Cyclone ROC Curves')
plot_roc_curves(Y_flood_test_bin, flood_preds_bin, 'Flood ROC Curves')
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.metrics import accuracy_score, classification_report


# Define Features & Targets
X = df.drop(columns=["CSI_Severity_Encoded", "flood_severity_numeric"])
y_cyclone = df["CSI_Severity_Encoded"]
y_flood = df["flood_severity_numeric"]

# 🚀 Train-Test Split (80% Train, 20% Test)
X_train, X_test, y_train_cyclone, y_test_cyclone, y_train_flood, y_test_flood = train_test_split(
    X, y_cyclone, y_flood, test_size=0.2, random_state=42
)

# 🚀 Step 1: Feature Selection using Random Forest
rf = RandomForestClassifier(n_estimators=200, random_state=42)
rf.fit(X_train, y_train_cyclone)

# Select Important Features
feature_importance = pd.Series(rf.feature_importances_, index=X.columns)
selected_features = feature_importance[feature_importance > 0.02].index.tolist()

# Reduce Features for Training
X_train_selected = X_train[selected_features]
X_test_selected = X_test[selected_features]

# 🚀 Step 2: Hyperparameter Tuning with Grid Search
xgb = XGBClassifier(random_state=42)
param_grid = {
    "n_estimators": [100, 200],
    "max_depth": [3, 5],
    "learning_rate": [0.01, 0.1]
}

# Grid Search for Cyclone Model
grid_search_cyclone = GridSearchCV(
    estimator=xgb,
    param_grid=param_grid,
    cv=3,
    scoring="accuracy",
    n_jobs=-1
)
grid_search_cyclone.fit(X_train_selected, y_train_cyclone)

# Grid Search for Flood Model
grid_search_flood = GridSearchCV(
    estimator=xgb,
    param_grid=param_grid,
    cv=3,
    scoring="accuracy",
    n_jobs=-1
)
grid_search_flood.fit(X_train_selected, y_train_flood)

# Best Parameters
best_params_cyclone = grid_search_cyclone.best_params_
best_params_flood = grid_search_flood.best_params_

print(f"✅ Best Parameters for Cyclone: {best_params_cyclone}")
print(f"✅ Best Parameters for Flood: {best_params_flood}")

# 🚀 Step 3: Train Best XGBoost Models
best_xgb_cyclone = XGBClassifier(**best_params_cyclone, random_state=42)
best_xgb_flood = XGBClassifier(**best_params_flood, random_state=42)

best_xgb_cyclone.fit(X_train_selected, y_train_cyclone)
best_xgb_flood.fit(X_train_selected, y_train_flood)

# 🚀 Step 4: Predictions
y_pred_cyclone = best_xgb_cyclone.predict(X_test_selected)
y_pred_flood = best_xgb_flood.predict(X_test_selected)

# 🚀 Step 5: Accuracy Scores
cyclone_train_acc = accuracy_score(y_train_cyclone, best_xgb_cyclone.predict(X_train_selected))
cyclone_test_acc = accuracy_score(y_test_cyclone, y_pred_cyclone)

flood_train_acc = accuracy_score(y_train_flood, best_xgb_flood.predict(X_train_selected))
flood_test_acc = accuracy_score(y_test_flood, y_pred_flood)

print(f"\n🚀 Cyclone Training Accuracy: {cyclone_train_acc:.4f}")
print(f"🚀 Cyclone Testing Accuracy: {cyclone_test_acc:.4f}")
print(f"🚀 Flood Training Accuracy: {flood_train_acc:.4f}")
print(f"🚀 Flood Testing Accuracy: {flood_test_acc:.4f}")

# 🚀 Step 6: Classification Reports
print("\n📊 Cyclone Severity Classification Report:\n")
print(classification_report(y_test_cyclone, y_pred_cyclone))

print("\n📊 Flood Severity Classification Report:\n")
print(classification_report(y_test_flood, y_pred_flood))
```

```python
import requests
import json
import numpy as np
import pandas as pd
import joblib
from datetime import datetime

# 🔑 OpenWeather API Key (Replace with your API Key)
API_KEY = "6969e1c7f7a0a29a0d3738e8477e2239"

# 🚀 Load Trained Models
best_cyclone = joblib.load("rf_xgb_cyclone.pkl")
best_flood = joblib.load("rf_xgb_flood.pkl")

# 🌍 Get Coordinates from City Name
def get_coordinates(city_name):
    geocode_url = f"http://api.openweathermap.org/geo/1.0/direct?q={city_name}&appid={API_KEY}"
    response = requests.get(geocode_url)
    if response.status_code == 200 and response.json():
        location = response.json()[0]
        return location["lat"], location["lon"]
    else:
        print("⚠️ Error: Could not fetch coordinates. Check city name.")
        return None, None

# 📥 Fetch Real-Time Weather Data
def get_weather_data(lat, lon):
    if lat is None or lon is None:
        return None
    weather_url = f"https://api.openweathermap.org/data/2.5/weather?lat={lat}&lon={lon}&appid={API_KEY}&units=metric"
    response = requests.get(weather_url)
    if response.status_code == 200:
        return response.json()
    print("⚠️ Error Fetching Weather Data:", response.status_code)
    return None

# 🛠️ Preprocess Weather Data for Model Input
def preprocess_data(weather_data):
    features = {
        "precip": weather_data.get("rain", {}).get("1h", 0),  # Rainfall in last 1hr
        "humidity": weather_data["main"]["humidity"] / 100,  # Normalize (0 to 1)
        "cloudcover": weather_data["clouds"]["all"] / 100,  # Normalize
        "windgust": weather_data.get("wind", {}).get("gust", 0),
        "windspeed": weather_data["wind"]["speed"],
        "sealevelpressure": weather_data["main"]["pressure"] / 1013,  # Normalize
        "STORM_SPEED": weather_data["wind"]["speed"] / 50,  # Approximate Max Wind
        "STORM_DIR": weather_data["wind"].get("deg", 0),
        "DIST2LAND": np.random.uniform(0, 100),  # Placeholder if unavailable
        "temp": weather_data["main"]["temp"],
        "feelslike": weather_data["main"]["feels_like"],
    }
    return pd.DataFrame([features])

# 🔧 Adjust Predictions Based on Confidence & Rainfall
def adjust_flood_prediction(probabilities, rainfall, threshold=0.95):
    max_prob = np.max(probabilities)
    pred_class = np.argmax(probabilities)
    if rainfall == 0 or max_prob < threshold:
        return 0  # Override to "No Flood"
    return pred_class

def adjust_cyclone_prediction(probabilities, windspeed, threshold=0.95):
    max_prob = np.max(probabilities)
    pred_class = np.argmax(probabilities)
    if windspeed < 10 or max_prob < threshold:
        return 0  # Override to "No Cyclone"
    return pred_class

# 🔮 Predict Cyclone & Flood Severity
def predict_severity(input_data):
    input_data_reshaped = input_data.values.reshape(1, 1, input_data.shape[1])  # Reshape for GRU model
    cyclone_probs = best_cyclone.predict_proba(input_data)
    flood_probs = best_flood.predict_proba(input_data)

    # Extract Key Features for Override Logic
    rainfall = input_data["precip"].values[0]
    windspeed = input_data["windspeed"].values[0]

    # Adjust Predictions
    cyclone_pred = adjust_cyclone_prediction(cyclone_probs, windspeed, threshold=0.95)
    flood_pred = adjust_flood_prediction(flood_probs, rainfall, threshold=0.95)

    # Debugging Logs
    print("\n🔍 Cyclone Prediction Probabilities:", cyclone_probs)
    print("🔍 Flood Prediction Probabilities:", flood_probs)

    # Convert Predictions to Labels
    cyclone_severity_map = {0: "No Cyclone", 1: "Low Cyclone", 2: "Moderate Cyclone", 3: "High Cyclone", 4: "Severe Cyclone"}
    flood_severity_map = {0: "No Flood", 1: "Low Flood", 2: "Moderate Flood", 3: "High Flood", 4: "Severe Flood"}

    return cyclone_severity_map[cyclone_pred], flood_severity_map[flood_pred]

# 🔄 Real-Time Prediction Process
def main():
    city_name = input("📌 Enter City Name: ").strip()

    print(f"\n🌍 Fetching Coordinates for {city_name}...\n")
    lat, lon = get_coordinates(city_name)
    if lat is None or lon is None:
        print("⚠️ Unable to fetch coordinates. Exiting.")
        return

    print(f"📌 Location Coordinates: {lat}, {lon}")

    print(f"\n🌍 Fetching Real-Time Weather Data for {city_name}...\n")
    weather_data = get_weather_data(lat, lon)

    if weather_data:
        print(f"📍 Requested Location: {city_name}")
        print(f"🌡️ Temperature: {weather_data['main']['temp']}°C")
        print(f"💨 Wind Speed: {weather_data['wind']['speed']} m/s")
        print(f"☁️ Cloud Cover: {weather_data['clouds']['all']}%")
        print(f"🌧️ Rainfall (Last 1hr): {weather_data.get('rain', {}).get('1h', 0)} mm")

        # 🛠️ Preprocess Data
        input_data = preprocess_data(weather_data)
        print("\n🛠️ Model Input Features:\n", input_data)

        # 🔮 Predict Cyclone & Flood Severity
        cyclone_result, flood_result = predict_severity(input_data)

        # 📝 Display Predictions
        print("\n🚨 **Real-Time Predictions:**")
        print(f"🌪️ Cyclone Severity: {cyclone_result}")
        print(f"🌊 Flood Severity: {flood_result}")
    else:
        print("⚠️ No weather data available.")

# 🔄 Run Real-Time Prediction
if __name__ == "__main__":
    main()
```
